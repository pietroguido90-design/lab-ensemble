```python
#Libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import AdaBoostClassifier
```

```python
spaceship = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/spaceship_titanic.csv")
spaceship.head()
```

```python
spaceship.info()
```

```python
spaceship.isnull().sum()
```

```python
spaceship_clean = spaceship.dropna()
```

**Perform Train Test Split**

```python
features = spaceship_clean.select_dtypes(include=np.number)
target = spaceship_clean["Transported"]            
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=17)
```

Now perform the same as before:
- Feature Scaling
- Feature Selection


```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
selector = SelectKBest(score_func=f_classif, k=5)

X_train_selected = selector.fit_transform(X_train_scaled, y_train)
X_test_selected = selector.transform(X_test_scaled)
```

**Model Selection** - now you will try to apply different ensemble methods in order to get a better model

- Bagging and Pasting

```python
bagging = BaggingClassifier(DecisionTreeClassifier(), n_estimators=100, bootstrap=True, random_state=17)
bagging.fit(X_train_scaled, y_train)
print("Bagging Accuracy:", bagging.score(X_test_scaled, y_test))
```

```python
pasting = BaggingClassifier(DecisionTreeClassifier(), n_estimators=100, bootstrap=False, random_state=17)
pasting.fit(X_train_scaled, y_train)
print("Pasting Accuracy:", pasting.score(X_test_scaled, y_test))
```

- Random Forests

```python
rf = RandomForestClassifier(n_estimators=100, random_state=17)
rf.fit(X_train_scaled, y_train)
print("Random Forest Accuracy:", rf.score(X_test_scaled, y_test))
```

- Gradient Boosting

```python
gb = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=17)
gb.fit(X_train_scaled, y_train)
print("Gradient Boosting Accuracy:", gb.score(X_test_scaled, y_test))
```

- Adaptive Boosting

```python
ada = AdaBoostClassifier(n_estimators=100, random_state=17)
ada.fit(X_train_scaled, y_train)
print("AdaBoost Accuracy:", ada.score(X_test_scaled, y_test))
```
